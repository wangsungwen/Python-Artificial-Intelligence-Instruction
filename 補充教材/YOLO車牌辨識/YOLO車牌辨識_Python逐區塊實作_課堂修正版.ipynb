## 可執行程式


```python
from pathlib import Path

def audit_rows(rows, class_count):
    issues = []
    for line_no, row in enumerate(rows, start=1):
        parts = row.split()
        if len(parts) != 5:
            issues.append((line_no, "欄位數不是 5"))
            continue
        try:
            class_id = int(parts[0])
            coords = [float(v) for v in parts[1:]]
        except ValueError as error:
            issues.append((line_no, f"數值格式錯誤: {error}"))
            continue
        if not 0 <= class_id < class_count:
            issues.append((line_no, "class 超界"))
        if any(v < 0 or v > 1 for v in coords):
            issues.append((line_no, "座標超界"))
    return issues

demo_rows = ["0 0.5 0.5 0.2 0.1", "0 1.2 0.5 0.2 0.1", "x 0.5 0.5"]
print(audit_rows(demo_rows, 1))
```

## 逐行語法解說

- **第 4 行 `issues=[]`**：建立容器，累積所有錯誤而非遇錯即停止。
- **第 5 行 `enumerate`**：同時取得人類可讀行號與內容。
- **第 6 行 `split`**：依空白把一行切成欄位。
- **第 7–9 行 `continue`**：欄位錯誤時跳到下一輪。
- **第 10–15 行 `try except`**：攔截 int/float 的 ValueError。
- **第 12 行 `list comprehension`**：把四個座標轉為 float。
- **第 16 行 `連鎖比較`**：等同 0 <= id and id < count。
- **第 18 行 `any`**：任一座標超界即成立。
- **第 20 行 `return`**：回傳完整問題清單。

## 可執行程式


```python
import re

def normalize_plate(text):
    return re.sub(r"[^A-Z0-9]", "", text.upper())

def valid_candidate(text):
    length_ok = bool(re.fullmatch(r"[A-Z0-9]{5,8}", text))
    has_digit = any(ch.isdigit() for ch in text)
    has_letter = any(ch.isalpha() for ch in text)
    return length_ok and has_digit and has_letter

samples = ["ab-c123", "1234", "ABCDEFG", "NY-3509"]
for sample in samples:
    text = normalize_plate(sample)
    print(sample, "->", text, valid_candidate(text))
```

## 逐行語法解說

- **第 3–4 行 `normalize_plate`**：先 upper，再用 re.sub 刪除非英數字元。
- **第 7 行 `fullmatch`**：整個字串必須是 5 到 8 個英數字。
- **第 8 行 `generator + any`**：至少一個字元 isdigit 為 True。

## 可執行程式


```python
import numpy as np

def order_characters(items, two_lines=False):
    if not items:
        return []
    if not two_lines or len(items) < 2:
        return sorted(items, key=lambda item: item[0])
    by_y = sorted(items, key=lambda item: item[1])
    gaps = [by_y[i + 1][1] - by_y[i][1]
            for i in range(len(by_y) - 1)]
    split = int(np.argmax(gaps)) + 1
    top = sorted(by_y[:split], key=lambda item: item[0])
    bottom = sorted(by_y[split:], key=lambda item: item[0])
    return top + bottom

demo = [(80, 20, 10, "C", .9), (20, 20, 10, "A", .9), (50, 20, 10, "B", .9)]
print("".join(item[3] for item in order_characters(demo)))
```

## 逐行語法解說

- **第 4 行 `not items`**：空 list 在布林情境為 False。
- **第 6–7 行 `sorted + lambda`**：以 item[0] 的 x 座標由左到右排序。
- **第 8 行 `item[1]`**：雙行時先依 y 座標由上到下排序。
- **第 9–10 行 `list comprehension`**：計算相鄰 y 差距。
- **第 11 行 `argmax`**：取得最大間距索引並加一成切割點。
- **第 12–13 行 `slice`**：切出上下兩排並各自依 x 排序。
- **第 14 行 `list + list`**：串接兩排的新清單。
- **第 17 行 `join`**：取 tuple 第四欄字元並合併。

## 可執行程式


```python
from PIL import Image, ImageDraw

def recognize_plates_bgr(image_bgr, plate_model, char_model):
    output = Image.fromarray(image_bgr[:, :, ::-1])
    draw = ImageDraw.Draw(output)
    height, width = image_bgr.shape[:2]
    result = plate_model.predict(image_bgr, conf=PLATE_CONF,
                                 device=DEVICE, verbose=False)[0]
    records = []
    if result.boxes is None:
        return np.array(output)[:, :, ::-1].copy(), records
    pairs = zip(result.boxes.xyxy.cpu().numpy(),
                result.boxes.conf.cpu().numpy())
    for box, score in pairs:
        x1, y1, x2, y2 = map(int, box)
        x1, y1 = max(0, x1), max(0, y1)
        x2, y2 = min(width, x2), min(height, y2)
        if x2 - x1 < 12 or y2 - y1 < 8:
            continue
        crop = image_bgr[y1:y2, x1:x2]
        record = recognize_one_crop(crop, char_model)
        record.update(box=(x1,y1,x2,y2), plate_conf=float(score))
        records.append(record)
        color = "limegreen" if record["accepted"] else "tomato"
        draw.rectangle((x1, y1, x2 - 1, y2 - 1), outline=color, width=2)
        draw.text((x1, max(0, y1 - 12)), record["plate"] or "?", fill=color)
    return np.array(output)[:, :, ::-1].copy(), records
```

## 逐行語法解說

- **第 2 行 `copy`**：建立可繪圖副本，不直接改輸入陣列。
- **第 3 行 `shape[:2]`**：取得 height、width 並拆包。
- **第 4–5 行 `predict()[0]`**：取得第一張影像的 Results。
- **第 6–8 行 `early return`**：沒有框就提早回傳空 records。
- **第 9–10 行 `zip`**：同步配對每個框與信心。
- **第 12 行 `map(int)`**：四座標轉成整數並拆包。

```python
def recognize_one_crop(crop, char_model):
    result = char_model.predict(crop, conf=CHAR_CONF,
                                device=DEVICE, verbose=False)[0]
    items = []
    if result.boxes is not None:
        arrays = zip(result.boxes.xyxy.cpu().numpy(),
                     result.boxes.conf.cpu().numpy(),
                     result.boxes.cls.cpu().numpy())
        for xyxy, confidence, class_id in arrays:
            x1, y1, x2, y2 = xyxy
            label = str(result.names[int(class_id)]).upper()
            if label == "-":
                continue
            items.append(((x1+x2)/2, (y1+y2)/2,
                          y2-y1, label, float(confidence)))
    ordered = order_characters(items)
    text = normalize_plate("".join(item[3] for item in ordered))
    weakest = min((item[4] for item in ordered), default=0.0)
    accepted = valid_candidate(text) and weakest >= TEXT_CONF
    return {"plate": text, "character_min_conf": weakest,
            "accepted": accepted}
```

## 可執行程式


```python
import shutil
import subprocess
from tempfile import TemporaryFile
from fractions import Fraction

SAMPLE_EVERY = 3
PREVIEW_VIDEO = False  # 在 Colab 將此改為 True，可在本格播放短片

def video_tools_ready():
    missing = [name for name in ("ffmpeg", "ffprobe") if shutil.which(name) is None]
    if missing:
        raise RuntimeError("影片單元需要 FFmpeg（Colab 通常已提供）：" + ", ".join(missing))

def video_metadata(path):
    video_tools_ready()
    try:
        probe = subprocess.run(
            ["ffprobe", "-v", "error", "-select_streams", "v:0", "-show_entries",
             "stream=width,height,avg_frame_rate", "-of", "json", str(path)],
            capture_output=True, text=True, check=True,
        )
        info = json.loads(probe.stdout)["streams"][0]
    except (subprocess.CalledProcessError, IndexError, KeyError) as exc:
        raise ValueError(f"無法開啟影片或找不到視訊軌：{path}") from exc
    width, height = int(info["width"]), int(info["height"])
    if width < 1 or height < 1:
        raise ValueError("影片尺寸不正確。")
    try:
        rate = Fraction(info["avg_frame_rate"])
        if not 1 <= rate <= 240:
            rate = Fraction(25)
    except (ValueError, ZeroDivisionError, KeyError):
        rate = Fraction(25)
    return width, height, str(rate)

def video_encoder(path, width, height, fps, stderr):
    return subprocess.Popen(
        ["ffmpeg", "-nostdin", "-v", "error", "-y", "-f", "rawvideo",
         "-pixel_format", "bgr24", "-video_size", f"{width}x{height}",
         "-framerate", str(fps), "-i", "pipe:0", "-an",
         "-vf", "pad=ceil(iw/2)*2:ceil(ih/2)*2", "-c:v", "libx264",
         "-pix_fmt", "yuv420p", str(path)],
        stdin=subprocess.PIPE, stderr=stderr,
    )

def create_demo_video(path):
    video_tools_ready()
    with TemporaryFile() as errors:
        with video_encoder(path, 320, 160, 6, errors) as writer:
            for index in range(9):
                writer.stdin.write(synthetic_frame(index).tobytes())
            writer.stdin.close()
            if writer.wait() != 0:
                errors.seek(0)
                raise RuntimeError("無法建立模擬影片：" + errors.read().decode(errors="replace"))

def process_video(source, output, plate_model, char_model, sample_every):
    if type(sample_every) is not int or sample_every < 1:
        raise ValueError("SAMPLE_EVERY 必須為正整數。")
    if source.resolve() == output.resolve():
        raise ValueError("輸入與輸出路徑不可相同。")
    width, height, fps = video_metadata(source)
    size = width * height * 3
    frame_index, sampled, accepted = 0, 0, 0
    stage = output.with_name(output.stem + ".partial.mp4")
    try:
        with TemporaryFile() as decode_log, TemporaryFile() as encode_log:
            with subprocess.Popen(
                ["ffmpeg", "-nostdin", "-v", "error", "-xerror", "-noautorotate",
                 "-i", str(source), "-map", "0:v:0", "-vsync", "0",
                 "-f", "rawvideo", "-pix_fmt", "bgr24", "pipe:1"],
                stdout=subprocess.PIPE, stderr=decode_log,
            ) as reader, video_encoder(stage, width, height, fps, encode_log) as writer:
                try:
                    while True:
                        data = reader.stdout.read(size)
                        if not data:
                            break
                        if len(data) != size:
                            raise ValueError("影片尾端含不完整影格。")
                        frame = np.frombuffer(data, dtype=np.uint8).reshape(height, width, 3)
                        frame_index += 1
                        if frame_index % sample_every == 0:
                            frame, records = recognize_plates_bgr(frame, plate_model, char_model)
                            sampled += 1
                            accepted += sum(record["accepted"] for record in records)
                        writer.stdin.write(frame.tobytes())
                    writer.stdin.close()
                    if reader.wait() != 0 or writer.wait() != 0 or not frame_index:
                        decode_log.seek(0)
                        encode_log.seek(0)
                        detail = (decode_log.read() + encode_log.read()).decode(errors="replace")
                        raise RuntimeError("影片轉換未完成或未讀到影格：" + detail)
                finally:
                    # 中途失敗先終止 FFmpeg，防止 subprocess 清理等待未讀完的 pipe。
                    if reader.poll() is None:
                        reader.kill()
                    if writer.poll() is None:
                        writer.kill()
        stage.replace(output)
    finally:
        if stage.exists():
            stage.unlink()
    return {"frames": frame_index, "sampled": sampled, "accepted": accepted}

VIDEO_RESULT = None
if USE_REAL_MODELS:
    # 已在第 10 單元載入模型，重新執行本格可更換影片。
    from google.colab import files
    videos = files.upload()
    if videos:
        name, data = next(iter(videos.items()))
        INPUT_VIDEO = PROJECT / ("real_input" + (Path(name).suffix or ".mp4"))
        INPUT_VIDEO.write_bytes(data)
        OUTPUT_VIDEO = PROJECT / "real_annotated.mp4"
    else:
        print("未選擇影片，已跳過。")
else:
    INPUT_VIDEO = PROJECT / "synthetic_input.mp4"
    OUTPUT_VIDEO = PROJECT / "synthetic_annotated.mp4"
    create_demo_video(INPUT_VIDEO)

if not USE_REAL_MODELS or videos:
    VIDEO_RESULT = process_video(INPUT_VIDEO, OUTPUT_VIDEO, plate_model, char_model, SAMPLE_EVERY)
    print(VIDEO_RESULT, "|", OUTPUT_VIDEO, "| 視覺輸出不含原音軌")
    if PREVIEW_VIDEO:
        from IPython.display import Video
        display(Video(str(OUTPUT_VIDEO), embed=True))
```

## 逐行語法解說（依程式區塊）

- `video_metadata()`：先確認檔案能開啟並讀取第一條視訊軌的尺寸與平均 FPS。
- `with ... as reader, ... as writer`：FFmpeg 程序負責影片解碼與編碼；`try/finally` 也涵蓋中途錯誤。
- `while True`：每次讀取一格 BGR 資料，讀到檔尾時 `break`。
- `frame_index += 1`：影格從 1 開始編號；`frame_index % sample_every == 0` 才推論。

- `stage.replace(output)`：完整成功才替換目標輸出；暫存檔最後清理。本教學版輸出不含音訊，依平均 FPS 產生等速影片，不承諾保留可變幀率時間戳。

## 可執行程式


```python
def edit_distance(a, b):
    previous = list(range(len(b) + 1))
    for i, char_a in enumerate(a, start=1):
        current = [i]
        for j, char_b in enumerate(b, start=1):
            insert_cost = current[-1] + 1
            delete_cost = previous[j] + 1
            replace_cost = previous[j-1] + (char_a != char_b)
            current.append(min(insert_cost, delete_cost, replace_cost))
        previous = current
    return previous[-1]

def evaluate_texts(truths, predictions):
    if len(truths) != len(predictions):
        raise ValueError("truths 與 predictions 數量不同")
    pairs = list(zip(truths, predictions))
    exact = sum(t == p for t, p in pairs) / max(1, len(pairs))
    errors = sum(edit_distance(t, p) for t, p in pairs)
    chars = sum(map(len, truths))
    cer = errors / max(1, chars)
    return {"exact_match": exact, "cer": cer}

print(evaluate_texts(["ABC123", "NY3509"], ["ABC128", "NY3509"]))
```

## 逐行語法解說

- **第 2 行 `初始列`**：建立 0 到 len(b) 的距離。
- **第 3–5 行 `巢狀迴圈`**：逐一比較兩字串的字元。
- **第 6–8 行 `三種成本`**：插入、刪除、替換。
- **第 8 行 `bool 算術`**：相同 False=0，不同 True=1。
- **第 9 行 `min`**：保留最低累積成本。
- **第 14–15 行 `raise`**：輸入筆數不同時主動報錯。
- **第 16 行 `zip`**：將真值與預測一一配對。
- **第 17 行 `sum(bool)`**：True 以 1 累加完全相同筆數。
- **第 18 行 `generator sum`**：加總每對編輯距離。
- **第 19 行 `map(len)`**：加總真值字元數。
- **第 20 行 `max(1,...)`**：避免除以零。
