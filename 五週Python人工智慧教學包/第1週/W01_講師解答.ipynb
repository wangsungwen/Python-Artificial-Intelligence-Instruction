**操作任務**：完成專題格；先測試35、3、200，再測試40、2、100。加上自己的品項。

```python
# 專題示範：更換這三個輸入即可重新計算
price, quantity, paid = 35, 3, 200
total = price * quantity
change = paid - total
print(f"紅茶 {quantity}杯，總價{total}元，找零{change}元")
```

### 練習
先在下方寫出預測，再取消程式註解並完成。

預測：＿＿＿＿

執行後發現：＿＿＿＿

```python
price, quantity, paid = 40, 2, 100
total = price * quantity
change = paid - total
assert (total, change) == (80, 20)
print(f"紅茶 {quantity}杯，總價{total}元，找零{change}元")
```

**參考答案與驗收**：測試A：105、95；測試B：80、20。輸出須含品項、杯數、總價、找零。
