### P078｜測試 ChatGPT

第一週講義第78頁

**講義文字（原稿轉錄）**

> 測試 ChatGPT

> 3n+1 猜想（考拉茲猜想）

> 對一個正整數 x 進行一連串運算：

> 如果 x 是奇數，則將其乘 3 再加 1；

> 而如果 x 是偶數，則將其除以 2，

> 如此重複運算最終都能夠得到 1。

#### 考拉茲猜想：可執行示範與邊界檢查

依原圖規則實作：奇數變3x+1、偶數整除2，至1停止。補充：對所有正整數都能抵達1仍是未證明的猜想，有限測試不是證明。加上步數上限避免任意輸入造成無限制執行；只接受正整數。

```python
def collatz_sequence(x, max_steps=10000):
    if type(x) is not int or x <= 0:
        raise ValueError('請輸入正整數')
    if type(max_steps) is not int or max_steps < 0:
        raise ValueError('步數上限須為非負整數')
    sequence = [x]
    steps = 0
    while x != 1:
        if steps >= max_steps:
            raise RuntimeError('已達步數上限，停止本次計算；這不是猜想的反例')
        x = x // 2 if x % 2 == 0 else 3 * x + 1
        sequence.append(x)
        steps += 1
    return sequence
ask = make_input(['6'])
x = int(ask('正整數 x：'))
print(collatz_sequence(x))
assert collatz_sequence(6) == [6,3,10,5,16,8,4,2,1]
assert collatz_sequence(1) == [1]
for invalid in [0, -1, 1.5, True]:
    try:
        collatz_sequence(invalid)
    except ValueError:
        print('已正確拒絕：', invalid)
    else:
        raise AssertionError('應拒絕非正整數')
try:
    collatz_sequence(3, max_steps=1)
except RuntimeError:
    print('步數上限檢查通過')
else:
    raise AssertionError('應觸發步數上限')
```

0、0.0、空字串、None、空容器通常為假；非空字串 "False"、"0" 都是真，不能用 bool 解析文字真假。str(1e32) 中的 + 是指數的正號，表示10的正32次方。

```python
# 我的練習區：請在此撰寫程式。
# 建議先寫輸入、處理、輸出三個步驟。
```

**參考解答與驗收**

觀察印出的結果，並閱讀 `assert` 檢核；若條件不成立，會出現 AssertionError。

```python
def order_total(raw_price, raw_quantity):
    price = float(raw_price)
    quantity = int(raw_quantity)
    if not math.isfinite(price) or price < 0 or quantity < 0:
        raise ValueError('單價與數量需為非負，且單價須為有限數值')
    return price * quantity
ask = make_input(['小安', '35.5', '2'])
customer_name = ask('姓名：')
raw_price = ask('單價：')
raw_quantity = ask('數量（整數）：')
total = order_total(raw_price, raw_quantity)
print(customer_name, '的總價為', total)
assert order_total('35.5', '2') == 71.0
assert order_total('35.5', '0') == 0.0
for price, quantity in [('-1','2'),('3','-1'),('nan','1'),('3','1.5')]:
    try:
        order_total(price, quantity)
    except ValueError:
        print('已拒絕不合法資料：', price, quantity)
    else:
        raise AssertionError('輸入驗證應拒絕此資料')
```
