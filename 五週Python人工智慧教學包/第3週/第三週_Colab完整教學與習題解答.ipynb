### P008｜範例

#### while 與終止條件

每次迴圈後更新 i。觀察離開迴圈時 i 已經超過 num；若一開始條件為假，區塊一次都不會執行。

```python
ask=make_input(['3'])
num=int(ask('請輸入整數：'))
i=1
while i<=num:
    print(f'i = {i}')
    i+=1
print(f'離開 while 迴圈，i = {i}')
```

### P009｜執行這程式會發生何事？

第三週講義第9頁

> 執行這程式會發生何事？

> 《Ctrl + c》可中止執行程式

#### 無窮迴圈的受控觀察

原碼 while i>=num 在 num<=1 時會一直成立；在 num=3 時則一次都不執行。以下加上最多五次的觀察上限，保留錯誤條件供除錯，不讓全部執行失去控制。

```python
num=1
i=1
steps=0
while i>=num and steps<5:
    print(i)
    i+=1
    steps+=1
print('五次後停止；原條件仍為',i>=num)
```

### P010｜浮點數的數值誤差

第三週講義第10頁

> 浮點數的數值誤差

> 0.01 + 0.02 + … + 0.99 + 1 = ?

> 正確答案是 50.5

#### 浮點迴圈的修正

用整數計數，再除以 100 產生小數。不要以反覆累加浮點數來判定是否剛好等於終點。

```python
i=1
total=0.0
while i<=100:
    total+=i/100
    i+=1
print(total)
assert math.isclose(total,50.5)
print('穩定加總：',math.fsum(i/100 for i in range(1,101)))
```

觀察印出的結果，並閱讀 `assert` 檢核；若條件不成立，會出現 AssertionError。

```python
def parse_number(text):
    text=text.strip()
    try:
        return int(text)
    except ValueError:
        value=float(text)
        if not math.isfinite(value):
            raise ValueError('須輸入有限數值')
        return value
ask=make_input(['69','-48','19','-91.3425','22.333'])
counts={'偶數':0,'奇數':0,'正數':0,'負數':0,'浮點數':0}
i=0
while i<5:
    value=parse_number(ask(f'第{i+1}個數：'))
    if type(value) is int:
        counts['偶數' if value%2==0 else '奇數']+=1
    else:
        counts['浮點數']+=1
    if value>0: counts['正數']+=1
    elif value<0: counts['負數']+=1
    i+=1
print(counts)
if not INTERACTIVE:
    assert counts=={'偶數':1,'奇數':2,'正數':3,'負數':2,'浮點數':2}
assert type(parse_number('0')) is int and type(parse_number('2.0')) is float
```

> 間隔值(step)：欲取出元素之間的間隔，若省略不給，則預設為 1，此時為串列的連續片段。

觀察印出的結果，並閱讀 `assert` 檢核；若條件不成立，會出現 AssertionError。

```python
def perfect_below(num):
    results=[]
    for n in range(2,num):
        divisors=[]
        for d in range(1,n//2+1):
            if n%d==0: divisors.append(d)
        if sum(divisors)==n: results.append(n)
    return results
ask=make_input(['1000'])
num=int(ask('正整數上限（建議<=10000）：'))
if num<1: raise ValueError('上限須為正整數')
print('完全數：',perfect_below(num))
assert perfect_below(30)==[6,28] and perfect_below(6)==[]
```

```python
def nearest_pair(points):
    if len(points)<2: return None
    best_distance=math.inf
    best_pair=None
    for i in range(len(points)):
        for j in range(i+1,len(points)):
            distance=math.hypot(points[i][0]-points[j][0],points[i][1]-points[j][1])
            if distance<best_distance:
                best_pair=(points[i],points[j])
                best_distance=distance
    return best_pair,best_distance
ask=make_input(['-3,-2','2,3','1,1','0,0'])
points=[]
while True:
    x,y=map(float,ask('點座標 x,y（0,0結束）：').split(','))
    if (x,y)==(0,0): break
    points.append((x,y))
print('最近的兩點與距離：',nearest_pair(points))
assert nearest_pair([]) is None
assert nearest_pair([(1,1),(1,1)])[1]==0
assert math.isclose(nearest_pair([(-3,-2),(2,3),(1,1)])[1],math.sqrt(5))
```

名稱查找可用 LEGB 理解：Local、Enclosing、Global、Built-in。函式內賦值預設建立區域變數；global 可重新綁定全域名稱。讀取全域或修改全域串列的內容不等於重新綁定名稱。

#### 習題解答｜直線與點的位置（第138–139頁）

斜率 a=(y1−y0)/(x1−x0)，截距 b=y0−a*x0；依題目回傳 [b,a]。比較 q_y 與 a*q_x+b，以容許誤差判斷線上。垂直線無法用 y=ax+b 表示，明確拒絕；補充一般式可處理垂直線。

**先練習**：依上方原題與提示，自行寫出步驟及程式。可修改輸入資料，測試正常值與邊界值。下方空白練習格不影響全部執行。

```python
# 我的練習區：請在此撰寫程式。
# 建議先寫輸入、處理、輸出三個步驟。
```

**參考解答與驗收**

觀察印出的結果，並閱讀 `assert` 檢核；若條件不成立，會出現 AssertionError。

```python
def lineEquation(x0,y0,x1,y1):
    if x0==x1:
        raise ValueError('重合點或垂直線無法用 y=ax+b 表示')
    a=(y1-y0)/(x1-x0)
    b=y0-a*x0
    return [b,a]
def position(lst,q_x,q_y):
    b,a=lst
    expected=a*q_x+b
    if math.isclose(q_y,expected,rel_tol=1e-9,abs_tol=1e-9): label='線上'
    elif q_y>expected: label='上面'
    else: label='下面'
    print(f'點 q({q_x:g},{q_y:g}) 在直線 L:y={a:g}x+{b:g} 的{label}')
    return label
ask=make_input(['1,5','3,9','4,2'])
x0,y0=map(float,ask('p0 的 x,y：').split(','))
x1,y1=map(float,ask('p1 的 x,y：').split(','))
qx,qy=map(float,ask('q 的 x,y：').split(','))
coef=lineEquation(x0,y0,x1,y1)
position(coef,qx,qy)
assert lineEquation(1,5,3,9)==[3,2]
assert position([3,2],4,2)=='下面'
assert position([3,2],2,10)=='上面'
assert position([3,2],-1,1)=='線上'
try:
    lineEquation(1,0,1,2)
except ValueError as e:
    print('邊界案例：',e)
# 補充：一般式 Ax+By+C=0 可以表示垂直線。
def general_line(x0,y0,x1,y1):
    if (x0,y0)==(x1,y1): raise ValueError('兩點不可重合')
    return y0-y1,x1-x0,x0*y1-x1*y0
print('垂直線的一般式係數：',general_line(1,0,1,2))
```
