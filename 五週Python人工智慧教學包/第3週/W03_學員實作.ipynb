## W3-07 摘要函數與空資料

090–110 分鐘，20分鐘。

- sum 與 len 求平均
- 先處理空串列
- 用字典回傳多個結果

**操作任務**：測試[59,60,81]與空串列；說明為何空資料平均不是0。

````python
def summarize(scores):
    if len(scores) == 0:
        return {"count": 0, "average": None, "passed": 0}
    total = 0
    passed = 0
    for score in scores:
        total += score
        if score >= 60:
            passed += 1
    return {"count": len(scores), "average": total / len(scores), "passed": passed}

print(summarize([59, 60, 81]))
print(summarize([]))
````

try:

except <Exception Type>:
    # 例外處理 
else:
    # 如果沒有例外，就會執行

    # 不論是否有產生例外，都會執行
上述語法的程式區塊說明，如下所示：

except程式區塊：當try程式區塊的程式碼丟出例外，我們需要準備一到多個except程式區塊來處理不同類型的例外。
else程式區塊：這是選項的程式區塊，可有可無，如果try程式區塊沒有產生例外，就會執行此程式區塊。
finally程式區塊：這是選項的程式區塊，可有可無，不論例外是否產生，都會執行此程式區塊的程式碼。
