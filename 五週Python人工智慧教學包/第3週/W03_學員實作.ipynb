## W3-07 摘要函數與空資料

090–110 分鐘，20分鐘。

- sum 與 len 求平均
- 先處理空串列
- 用字典回傳多個結果

**操作任務**：測試[59,60,81]與空串列；說明為何空資料平均不是0。

```python
def summarize(scores):
    if len(scores) == 0:
        return {"count": 0, "average": None, "passed": 0}
    total = 0
    passed = 0
    for score in scores:
        total += score
        if score >= 60:
            passed += 1
    return {"count": len(scores), "average": total / len(scores), "passed": passed}

print(summarize([59, 60, 81]))
print(summarize([]))
```

## W3-09 例外與輸入檢查

125–140 分鐘，15分鐘。

- try 嘗試可能失敗的轉型
- except ValueError 處理格式錯誤
- 範圍不合法也需要拒絕

**操作任務**：測試"60"、"abc"、"120"、"-1"。

```python
def parse_score(raw):
    try:
        value = int(raw)
    except ValueError:
        return None
    if value < 0 or value > 100:
        return None
    return value

for raw in ["60", "abc", "120", "-1"]:
    print(raw, parse_score(raw))
```
