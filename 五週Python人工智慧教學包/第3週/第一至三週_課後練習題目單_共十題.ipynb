def bmi_cm(height, weight):
    if not (math.isfinite(height) and math.isfinite(weight)) or height <= 0 or weight <= 0:

    return weight / (height / 100) ** 2

def 溫度轉換(unit, temp):
    """unit 為輸入溫度單位：1=華氏，2=攝氏；回傳轉換後的溫度。"""
    if unit == 1:
        return (temp - 32) * 5 / 9
    elif unit == 2:
        return temp * 9 / 5 + 32
    else:
        raise ValueError('輸入單位只能為 1 或 2')

# 輸入：讀取單位與溫度
try:
    unit = int(input('輸入單位 1=華氏轉攝氏、2=攝氏轉華氏：'))
    temp = float(input('溫度：'))

    # 處理：傳遞兩個參數，接收函式回傳值
    result = 溫度轉換(unit, temp)

    # 輸出：在函式外顯示結果
    target_unit = '攝氏' if unit == 1 else '華氏'
    print(f'轉換結果：{result:.2f} 度{target_unit}')
except ValueError as error:
    print('輸入錯誤：', error)

# 驗收：檢查兩個轉換方向與負溫度
assert abs(溫度轉換(1, 32) - 0) < 1e-9
