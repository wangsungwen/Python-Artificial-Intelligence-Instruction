## W2-08 range 與 while

110–125 分鐘，15分鐘。

- range(1,4) 產生1、2、3
- while 在條件成立時繼續
- 必須讓條件有機會變成假

**操作任務**：將倒數改成5到1；說明拿掉remaining -= 1會發生什麼。

```python
for i in range(1, 4):
    print("第", i, "次")
remaining = 3
while remaining > 0:
    print(remaining)
    remaining -= 1
```

## W2-09 迴圈內的條件

125–140 分鐘，15分鐘。

- 每筆訂單各自決定運費
- subtotal 與 grand_total 分開
- 門檻要以商品金額判斷

**操作任務**：計算[80,100,120]每筆付款與總付款，畫出三次追蹤表。

```python
amounts = [80, 100, 120]
grand_total = 0
for amount in amounts:
    if amount >= 100:
        shipping = 0
    else:
        shipping = 20
    subtotal = amount + shipping
    grand_total += subtotal
    print(amount, shipping, subtotal)
print("總付款", grand_total)
```
