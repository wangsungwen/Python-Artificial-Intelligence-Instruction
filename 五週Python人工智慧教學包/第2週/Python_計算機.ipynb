## C1 運算式與運算子

**講授：** 運算式會產生一個值，`7 + 3`的兩個運算元是7和3，運算子是`+`。`total = 7 + 3`先算右邊，再將結果交給名稱`total`。按鍵文字是字串，`"7" + "3"`會串接成`"73"`；轉為`float`後才是數值加法。`input()`回傳的也是字串，輸入小數時用`float`，不要把`"3.5"`直接交給`int`。

| 運算子 | 用途 | 例子與結果 |
|---|---|---|
| `+ - * /` | 加、減、乘、真除法 | `7/2`為3.5 |
| `// %` | 向下取整除法、餘數 | `7//2`為3；`7%2`為1 |
| `**` | 次方 | `2**3`為8 |
| `+=` | 計算後重新指派 | 字串可累加按鍵，數值可累加結果 |
| `== != < <= > >=` | 比較大小或相等 | 結果為True或False |
| `and or not` | 組合條件、反轉真假 | 先檢查非零再進行除法 |
| `in / not in` | 成員檢查 | `op in ["/", "//", "%"]` |

常用優先順序可先記「括號、次方、乘除餘數、加減、比較、not、and、or」。負號和次方要特別看括號：`-2**2`是-4，`(-2)**2`是4；次方由右往左結合。`a/b`、`a//b`、`a%b`的除數都不能為0。用括號表達計算意圖比背完整表格更可靠。

浮點數保存近似值，`0.1+0.2`可能顯示長小數。顯示兩位可用`f"{result:.2f}"`，這不會改變內部數值；測試近似值可用`math.isclose`。位元運算`& | ^ ~ << >>`是延伸項，`^`是位元互斥或，不能當作次方。

**預測與檢核：** 先寫出以下每個輸出，再執行。請說明`-7//2`為何不是-3。

````python
import math
print("字串串接：", "7" + "3")
print("數值加法：", float("7") + float("3"))
print("先乘後加：", 2 + 3 * 4, "括號優先：", (2 + 3) * 4)
print("除法、整除、餘數：", 7 / 2, -7 // 2, -7 % 2)
print("次方：", 2 ** 3, -2 ** 2, (-2) ** 2)
print("浮點結果：", 0.1 + 0.2, math.isclose(0.1 + 0.2, 0.3))
print("格式化：", f"{7 / 3:.2f}")
print("位元延伸：", 5 & 3, 5 | 3, 5 ^ 3)
````

## C2 判斷式與按鍵流程

**講授：** `if`只在條件成立時執行縮排區塊；`else`承接不成立的情況。多種互斥動作使用`if/elif/else`，由上到下只選第一個成立的分支。每行條件後加冒號，區塊用四個空白，避免混用Tab。

按鍵流程是：AC清空；等號進行計算；其他按鍵追加字串。若目前顯示Error，下一個數字應開始新的輸入。這是外層選按鍵、內層檢查狀態的巢狀判斷。條件式運算`"0" if text == "" else text`產生一個顯示值，適合很短的二選一。

`and`會在左邊是假時停止檢查：`b != 0 and a/b > 2`可避免在b為0時做除法。`&`會計算兩邊，不能用來替換這個防護。

**逐步追蹤：** 下例輸入`7,+,3,AC,9`，每輪依序得到`7`、`7+`、`7+3`、空字串、`9`。這裡遇到等號只作流程標記，真正計算於C3及C7補齊。

````python
expression = ""
for key in ["7", "+", "3", "AC", "9"]:
    if key == "AC":
        expression = ""
    elif key == "=":
        print("交給計算函式")
    else:
        if expression == "Error":
            expression = ""
        expression += key
    print(key, "=>", expression if expression != "" else "0")
a, b = 10, 0
print("短路檢查：", b != 0 and a / b > 2)
````

## C3 函式把運算規則集中管理

**講授：** `def calculate(a, op, b):`只定義方法，呼叫`calculate(7, "+", 3)`才執行。定義中的a、op、b是參數，呼叫時的7、`"+"`、3是引數。函式內的名稱原則上是區域變數，外面的介面只需拿回傳值，不必知道內部用了哪個分支。

`return`把答案交回呼叫端，並立刻結束函式；`print`只負責顯示。若函式只有print而沒有return，取得的結果會是None。把「計算」和「顯示」拆開，才能重複用於Notebook、網頁按鈕與Tkinter。

下面先檢查輸入為有限數值，接著排除除以零，最後選擇運算分支。錯誤以`raise ValueError`交給上層處理，不把`"Error"`混進數值答案。`try/except`是錯誤處理延伸語法，用來讓單筆錯誤不終止整個示範。

**檢核：** `calculate(8,"/",2)`回傳4.0；`calculate(8,"/",0)`提出錯誤。試著把呼叫結果再乘以2，體會return的用途。

````python
def calculate(a, op, b):
    if type(a) not in (int, float) or type(b) not in (int, float):
        raise ValueError("運算元必須是數值")
    if not math.isfinite(a) or not math.isfinite(b):
        raise ValueError("請輸入有限數值")
    if op in ["/", "//", "%"] and b == 0:
        raise ValueError("除數不可為0")
    if op == "+":
        return a + b
    elif op == "-":
        return a - b
    elif op == "*":
        return a * b
    elif op == "/":
        return a / b
    elif op == "//":
        return a // b
    elif op == "%":
        return a % b
    elif op == "**":
        if abs(a) > 1000000 or abs(b) > 10:
            raise ValueError("本教學版次方限定底數絕對值不超過1000000，指數不超過10")
        result = a ** b
        if isinstance(result, complex):
            raise ValueError("本教學版僅處理實數")
        return result
    raise ValueError("不支援的運算子")

print(calculate(8, "/", 2) * 2)
try:
    calculate(8, "/", 0)
except ValueError as error:
    print("已處理：", error)
````

## C5 for與巢狀迴圈

**講授：** `for value in data`依序取得每個元素，處理已知的一批資料很方便。`range(4)`產生0、1、2、3，不含4；`range(1,5)`是1至4；`range(5,0,-1)`是5至1，間隔不能為0。想同時取得索引與值可用`enumerate`。

按鈕的外層迴圈每次取一列，內層每次取一個字元。原範例在每列開始時重新建立`row_widgets=[]`，避免把前面列的按鈕重複放進來。6列、每列4個按鍵，共建立24個按鈕；不是6+4個。

第二段以for批次處理運算資料。遇錯用continue只略過該次，下一筆仍可執行。迴圈後的else是在未被break中止時執行，與if的else意義不同。巢狀迴圈中的break只跳出最近的一層。

````python
print("range示範：", list(range(4)), list(range(1, 5)), list(range(5, 0, -1)))
button_count = 0
for row_index, row_data in enumerate(buttons_data):
    for column_index, key in enumerate(row_data):
        button_count += 1
        print(f"({row_index},{column_index})={key}", end="  ")
    print()
print("按鈕總數：", button_count)

jobs = [(7, "+", 3), (9, "/", 0), (8, "%", 3)]
batch_results = []
for a, op, b in jobs:
    try:
        result = calculate(a, op, b)
    except ValueError as error:
        print("略過：", error)
        continue
    batch_results.append(result)
else:
    print("所有工作已走訪完成：", batch_results)
````

## C6 while、停止條件與互動輸入

**講授：** while在每次執行區塊前檢查條件，成立才執行；如果一開始就是False，會執行零次。設計時找出三件事：初值、持續條件、狀態更新。下例初值index=0，持續條件`index<len(commands)`，每輪執行`index+=1`。

我們先以串列模擬使用者操作，讓整份筆記本可自動跑完。空指令用continue略過；quit用break離開。index必須在continue之前增加，否則可能永遠讀到同一個空字串。真正的命令列輸入通常使用while True，讀到q便break；原GUI則由事件迴圈等待點擊，不要在按鍵回呼內再寫阻塞的while True。

**檢核：** 下例應只記錄7與3；quit之後的9不會被處理。把quit移除後，迴圈仍會在清單結尾停止。

````python
commands = ["7", "", "3", "quit", "9"]
index = 0
accepted = []
while index < len(commands):
    command = commands[index]
    index += 1
    if command == "":
        continue
    if command == "quit":
        break
    accepted.append(command)
print("收到的數字：", accepted)

def run_console_calculator():
    while True:
        raw = input("第一個數（q離開）：").strip()
        if raw.lower() == "q":
            break
        try:
            a = float(raw)
            op = input("運算子：").strip()
            b = float(input("第二個數："))
            print("答案：", calculate(a, op, b))
        except (ValueError, ZeroDivisionError, OverflowError) as error:
            print("請重新輸入：", error)

RUN_INPUT_DEMO = False  # 想體驗input時改成True；輸入q結束
if RUN_INPUT_DEMO:
    run_console_calculator()
````

## C7 完整算式與原eval的對照

**講授：** C3處理一個二元運算，原按鈕計算機卻能累積`2+3*4`。完整算式需要辨識先後順序，不能一律從左算到右。原碼使用`eval`直接求值；投影片Py01(2)第34–37頁正好說明此功能與風險。

以下是選讀的解析工具：用Python的ast讀取算式結構，只接受數字、正負號及七種算術運算，運算時仍呼叫C3的calculate。這樣按鈕及輸入欄位都不會呼叫任意名稱或函式。字元長度、節點數及次方範圍也有限制。初學者本週須理解「輸入字串、回傳數值、錯誤可捕捉」，遞迴解析留作進階閱讀。

**對照：** 原`current_expression.replace('x','*')`仍保留；原`eval(eval_expr)`改成`evaluate_expression`。這是實作調整，不是原PPT的程式。

````python
import ast
import html

AST_OPERATORS = {ast.Add: "+", ast.Sub: "-", ast.Mult: "*", ast.Div: "/",
                 ast.FloorDiv: "//", ast.Mod: "%", ast.Pow: "**"}

def evaluate_expression(text):
    expression = text.strip().replace("x", "*")
    if not expression or len(expression) > 100:
        raise ValueError("請輸入1至100字的算式")
    tree = ast.parse(expression, mode="eval")
    if sum(1 for _ in ast.walk(tree)) > 80:
        raise ValueError("算式過長")
    def visit(node):
        if isinstance(node, ast.Constant) and type(node.value) in (int, float):
            value = node.value
        elif isinstance(node, ast.UnaryOp) and isinstance(node.op, (ast.UAdd, ast.USub)):
            value = visit(node.operand)
            if isinstance(node.op, ast.USub):
                value = -value
        elif isinstance(node, ast.BinOp) and type(node.op) in AST_OPERATORS:
            value = calculate(visit(node.left), AST_OPERATORS[type(node.op)], visit(node.right))
        else:
            raise ValueError("只接受數字與算術運算")
        if not math.isfinite(value) or abs(value) > 1e100:
            raise ValueError("結果超出教學版範圍")
        return value
    return visit(tree.body)

for expression in ["2+3*4", "(2+3)*4", "-7//2", "0.5+1.25", "2**3"]:
    print(expression, "=", evaluate_expression(expression))
````

## C8 整合網頁按鈕與事件回呼

**講授：** 先追蹤一次按鍵生命週期：建立Button、用`on_click(on_button_click)`登記處理函式、使用者點擊、系統將按鈕物件b傳入、讀取b.description、更新算式、更新HTML。登記時不加`()`，因為此時只是交出函式，尚未要執行它。

此處保留原範例的全域字串做入門說明。函式內重新指派current_expression時需宣告global；C3的數值計算函式沒有改全域狀態，所以較容易測試。新增DEL刪去末字元；±將目前整段算式取負，並非只改最後一個運算元。

for迴圈只負責建立24個按鈕；互動之後交給Jupyter的事件機制，不必額外使用while。操作測試：`AC,2,+,3,x,4,=`應得14；`AC,8,/,0,=`應顯示Error；之後按7應重新開始。

````python
try:
    import ipywidgets as widgets
    from IPython.display import display
except ImportError:
    widgets = None
    print("互動介面需ipywidgets；基本語法與測試仍可執行。安裝指令：%pip install ipywidgets")

current_expression = ""
html_display = widgets.HTML() if widgets is not None else None

def update_display(text):
    display_text = text if text != "" else "0"
    if html_display is not None:
        html_display.value = ('<div style="padding:12px;border:1px solid #777;'
                              'font:24px monospace;text-align:right;overflow-wrap:anywhere">'
                              + html.escape(display_text) + '</div>')

def on_button_click(b):
    global current_expression
    key = b.description
    if key == "AC":
        current_expression = ""
    elif key == "=":
        try:
            current_expression = str(evaluate_expression(current_expression))
        except (ValueError, SyntaxError, ZeroDivisionError, OverflowError, RecursionError):
            current_expression = "Error"
    elif key == "DEL":
        current_expression = "" if current_expression == "Error" else current_expression[:-1]
    elif key == "±":
        if current_expression and current_expression != "Error":
            current_expression = f"-({current_expression})"
    else:
        if current_expression == "Error":
            current_expression = ""
        if len(current_expression) + len(key) <= 100:
            current_expression += key
    update_display(current_expression)

if widgets is not None:
    rows = []
    for row_data in buttons_data:
        row_widgets = []
        for key in row_data:
            button = widgets.Button(description=key, layout=widgets.Layout(width="60px", height="42px"))
            if key in ["+", "-", "x", "/", "//", "%", "**", "="]:
                button.style.button_color = "#f6c56f"
            button.on_click(on_button_click)
            row_widgets.append(button)
        rows.append(widgets.HBox(row_widgets))
    update_display("")
    calculator_ui = widgets.VBox([html_display] + rows)
    display(calculator_ui)
````

## C11 本機Tkinter選讀

此完整桌面版採「兩個數值加運算子」介面，讓初學者專注C3核心函式。將C1的`import math`、C3的`calculate`函式及下列程式一起存為.py，在支援Tkinter的本機Python執行。Colab不適用；程式區塊以文字呈現，不會隨全部執行啟動視窗。

```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
root.title("兩數計算機")
left = ttk.Entry(root)
left.grid(row=0, column=0, padx=8, pady=8)
operator = ttk.Combobox(root, values=["+", "-", "*", "/", "//", "%", "**"], state="readonly", width=5)
operator.set("+")
operator.grid(row=0, column=1)
right = ttk.Entry(root)
right.grid(row=0, column=2, padx=8, pady=8)
answer = tk.StringVar(value="請輸入兩個數值")

def compute():
    try:
        value = calculate(float(left.get()), operator.get(), float(right.get()))
        answer.set(str(value))
    except (ValueError, ZeroDivisionError, OverflowError) as error:
        answer.set(str(error))

ttk.Button(root, text="計算", command=compute).grid(row=1, column=0, columnspan=3)
ttk.Label(root, textvariable=answer).grid(row=2, column=0, columnspan=3, pady=8)
root.mainloop()
```

`mainloop()`由視窗系統處理事件；`command=compute`登記函式而不是當場呼叫。這與Notebook的`on_click`概念相同。
