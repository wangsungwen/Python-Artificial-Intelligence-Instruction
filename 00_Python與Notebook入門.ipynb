## 1 文字與程式
Markdown 格負責說明；Code 格讓電腦執行指令。`print` 把訊息顯示出來，文字要放在引號內。預期看到「我的第一個 AI 實驗」。

```python
print("我的第一個 AI 實驗")
```

## 2 變數與參數
變數像有名字的盒子。等號是把右邊存進左邊；`#` 後面是給人看的註解。請把 epochs 改成 5，再執行一次。

```python
epochs = 3  # 訓練資料重複閱讀次數
batch = 4   # 一次處理幾張圖
conf = 0.5  # 偵測分數門檻，不是正確率保證
print("epochs:", epochs, "batch:", batch, "conf:", conf)
```

## 3 清單、字典與索引
清單按順序保存項目，第一項的索引是 0。字典用名稱找資料。

```python
classes = ["red", "yellow", "green"]
settings = {"epochs": 3, "imgsz": 320}
print(classes[0])
print(settings["epochs"])
```

## 4 迴圈與條件
for 讓相同動作重複；if 依條件選擇動作。冒號後面的縮排是程式的一部分。預期只通過 0.7 與 0.9。

```python
scores = [0.2, 0.7, 0.9]
for score in scores:
    if score >= conf:
        print("保留", score)
```

## 5 函式
函式把一組步驟命名，以後只要呼叫名稱。參數是輸入，return 是回傳結果。

```python
def precision(tp, fp):
    return tp / (tp + fp) if tp + fp else 0.0

print(precision(8, 2))
assert precision(8, 2) == 0.8
```

## 6 路徑與檔案
Path 表示檔案位置。工作目錄會影響相對路徑；副檔名不同通常用途不同。

```python
from pathlib import Path
print("工作目錄:", Path.cwd())
for name in ["data.yaml", "best.pt", "test.jpg", "results.csv"]:
    print(name, "副檔名:", Path(name).suffix)
```

## 7 評估指標
TP 正確偵測、FP 誤報、FN 漏報。以下是人工指定的教學數字，不是模型實驗結果。

```python
tp, fp, fn = 8, 2, 4
p = tp / (tp + fp)
r = tp / (tp + fn)
f1 = 2*p*r/(p+r)
print(f"Precision={p:.3f}, Recall={r:.3f}, F1={f1:.3f}")
```

## 8 自己完成
把 TP、FP、FN 改成 6、3、2，先在紙上預測 Precision 與 Recall，再執行檢查。

```python
tp, fp, fn = 6, 3, 2
print("Precision:", tp/(tp+fp))
print("Recall:", tp/(tp+fn))
```

## 常見問題
- NameError：先執行建立該變數的上方程式格。
- ModuleNotFoundError：先執行安裝格，再重新匯入。
- FileNotFoundError：確認檔名、大小寫與目前工作目錄。
- CUDA out of memory：降低 batch 或 imgsz，再重新啟動執行階段。
- 安裝格重新執行後仍有版本錯誤：重新啟動，再從匯入格開始。
